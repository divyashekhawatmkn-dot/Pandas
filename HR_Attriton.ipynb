```python
#Task 50: Create an AgeGroup column:
#         Age < 30       = Young
 #        Age 30-39      = Adult
  #       Age 40-49      = Middle_Age
    #Age >= 50      = Senior

import numpy as np


conditions = [
    df["Age"] < 30,
    (df["Age"] >= 30) & (df["Age"] <= 39),
    (df["Age"] >= 40) & (df["Age"] <= 49),
    df["Age"] >= 50
]

choice =[
    "Young",
    "Adult",
    "Middle_Age",
    "Senior"
]



df["Age_group"] = np.select(conditions,choice,default="Unknown")
print(df[["Age","Age_group"]])

    
```

```python
#Task 51: Create an IncomeLevel column using MonthlyIncome:
#         Low
 #        Medium
  #       High

conditions = [
    df["MonthlyIncome"] < 5000,
    (df["MonthlyIncome"] >= 5000) & (df["MonthlyIncome"] <= 15000),
    df["MonthlyIncome"] >= 19000
]

choice = [
    "LOW",
    "MEDIUM",
    "HIGH"
]


df["Income_level"] = np.select(conditions,choice, default = "unknown")
print(df[["MonthlyIncome","Income_level"]])
    

```

```python
#Task 52: Create an ExperienceLevel column using TotalWorkingYears:
#         Beginner
 #        Intermediate
  #       Experienced
   #      Highly Experienced


conds = [
    df["TotalWorkingYears"] < 4,
    (df["TotalWorkingYears"] >= 4) & (df["TotalWorkingYears"] <= 15),
    (df["TotalWorkingYears"] >= 16) & (df["TotalWorkingYears"] <= 25),
    df["TotalWorkingYears"] > 25
]

dekh = [
    "Beginner",
    "Intermediate",
    "Experienced",
    "Highly Experienced"
]

df["ExperienceLevel"] = np.select(conds,dekh,default = "unknown")
print(df[["TotalWorkingYears","ExperienceLevel"]])
```

```python
#ask 53: Create a TenureLevel column using YearsAtCompany:
#         New
#         Mid Tenure
 #        Long Tenure


conds =[
    df["YearsAtCompany"] < 5,
    (df["YearsAtCompany"] >=5) & (df["YearsAtCompany"] <= 17),
    df["YearsAtCompany"] > 17
]

choice = [
    "NEW",
    "MID TENURE",
    "LONG TENURE"
]

df["TenureLevel"] = np.select(conds,choice,default ="unknown")
print(df[["YearsAtCompany","TenureLevel"]])
```

```python
#ask 54: Create a PromotionStatus column using YearsSinceLastPromotion:
 #        Recently Promoted
#         Not Recently Promoted


conditions = [
    df["YearsSinceLastPromotion"] < 6,
    df["YearsSinceLastPromotion"] >= 6
]

dekh =[
    "Recently_Promoted",
    "Not_Recently_Promoted"
]

df["PromotionStatus"] = np.select(conditions,dekh,default="unknown")
print(df[["YearsSinceLastPromotion","PromotionStatus"]])
```

```python
#Task 55: Create a WorkLifeCategory column:
 #        Poor
  #       Good
#      Excellent



conditions = [
    df["WorkLifeBalance"] == 1,
    df["WorkLifeBalance"] == 2,
    df["WorkLifeBalance"] >= 3
]

choices = [
    "Poor",
    "Good",
    "Excellent"
]

df["WorkLifeCategory"] = np.select(conditions, choices, default = "unknown")

print(df[["WorkLifeBalance", "WorkLifeCategory"]])

```

```python
#Task 56: Create a SatisfactionScore using:
#         EnvironmentSatisfaction
#         JobSatisfaction
#         RelationshipSatisfaction
#         WorkLifeBalance


df["SatisfactionScore"] =(
    df["EnvironmentSatisfaction"]
    + df["JobSatisfaction"]
    + df["RelationshipSatisfaction"]
    + df["WorkLifeBalance"]
)

print(df[[
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "SatisfactionScore"
]])
```

```python
#Task 57: Create a SatisfactionLevel column:
#         Low
#         Medium
#         High

a = [
    df["SatisfactionScore"] < 5,
    (df["SatisfactionScore"] >= 5) & (df["SatisfactionScore"] <= 10),
    df["SatisfactionScore"] <= 16

]

b = [
    "LOW",
    "MEDIUM",
    "HIGH"
]

df["SatisfactionLevel"] = np.select(a,b,default = "unknown")

print(df[["SatisfactionScore","SatisfactionLevel"]])
```
