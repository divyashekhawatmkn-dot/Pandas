```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Hp\Downloads\Untitled.csv")
print(df)
```

```python
#Data Cleaning
#1. find missing value
print(df.isnull().sum())
```

```python
#2.Remove Duplicate record
print(df.duplicated())
print(df.drop_duplicates())
```

```python
#3.Convert date columns into datetime format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
print(df)
```

```python
#4.Rename Columns
print(df.rename(columns = {"Customer_Name":"Name"}))
```

```python
#5. Check Datatype
print(df.dtypes)
#6.There is no missing value
```

```python
#7.Standardize product name
df = df["Product"].str.title()
print(df)
```

```python
print(df)
```

```python
#Phase 3: Data Analysis
#1.Total number of order

df = df["Order_ID"].count()
print(df)
```

```python
#2.Total revenue
print(df["Unit_Price"].sum())
```

```python
#3.Average Order Value
df["AOV"] = df["Unit_Price"].mean()
print(df)
```

```python
#4.Top 10 Product
top_10 = (
    df.groupby("Product")["Quantity"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
print(top_10)
```

```python
#5. bottom 10
bottom_10 = (
    df.groupby("Product")["Quantity"]
      .sum()
      .sort_values(ascending=False)
      .tail(10)
)
print(bottom_10)
```

```python
#6.City Wise Sale
df = df.groupby("City")["Unit_Price"].sum()
print(df)
```

```python
#7.Category wise Sale
df =df.groupby("Category")["Unit_Price"].sum()
print(df)
```

```python
#8.Most profitabe City
df = df.groupby("Unit_Price")["City"].sum().max()
print(df)
```

```python
#9.Highest rated product
df = df["Rating"].max()
print(df)
```

```python
#10.Most used payment method
df = df["Payment_Method"].max()
print(df)
```

```python
#Phase :4 Customer Analytics

#1. Total customer
df = df["Customer_ID"].count()
print(df)
```

```python
#2.Repeated customer
co = df.groupby("Customer_ID")["Order_ID"].nunique()
repeat_customer = co[co > 1]

print(repeat_customer)
```

```python
#3.One time customer
one_time = co[co == 1]
print(one_time)
```

```python
#4. find top customer
df = df.groupby("Customer_ID")["Unit_Price"].sum().nlargest(5)
print(df)
```

```python
#5.Customer Life Time value

clv = df.groupby("Customer_ID")["Unit_Price"].sum()
print(clv.sort_values(ascending = False).head(10))
```

```python
#Retention Rate
repeat_count = len(co)
total_count = df["Customer_ID"].nunique()
retention_rate=(repeat_count /total_count) * 100
print("Retention Rate:",retention_rate)
```

```python
#7.Purchasing patterns

pp = df.groupby("Customer_ID").agg(
    Total_orders=("Order_ID","nunique"),
    Total_spending=("Unit_Price","sum"),
    AVG_spending=("Unit_Price","mean")
)

print(pp)
```

```python
#8.Average spending
avg_sp = df.groupby("Customer_ID")["Unit_Price"].mean()
print("Average Spending:",avg_sp)
```

```python
#9.Monthly Active Customer
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.to_period("M")
print(df)

monthly_customer = df.groupby("Month")["Customer_ID"].nunique()
print("\nMonthly Active Customer:\n",monthly_customer)
```

```python
# 10.Customer Report
customer_report = df.groupby("Customer_ID").agg(
    Total_Orders=("Order_ID","nunique"),
    Total_Spending=("Unit_Price","sum"),
    Average_Spending=("Unit_Price","mean"),
    Average_Rating=("Rating", "mean"),
    City=("City", "first")
)
customer_report["Customer_Type"] = customer_report["Total_Orders"].apply(
    lambda x:"Repeat Customer" if x > 1 else"One-Time Customer"
)
print(customer_report)
```

```python
#Phase:5 Product Analytics

#1.Best selling product
df = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head(10)
print(df)
```

```python
#Least selling product
df = df.groupby("Product")["Quantity"].sum().sort_values().head(10)
print(df)
```

```python
#3.Product rating
df = df.groupby("Product")["Rating"].mean().sort_values(ascending=False).head(10)
print(df)
```

```python
#Category growth
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.to_period("M")

growth = (df.groupby(["Month","Category"])["Unit_Price"].sum().unstack())

print(growth)
```

```python
#Stock Requriements

md = df.groupby(["Month","Product"])["Quantity"].sum()

stock = md.groupby("Product").mean().head(10)

print(stock)
```

```python
#8.Product Trends
pt = df.groupby(["Month","Product"])["Unit_Price"].sum().unstack()
print(pt)
```

```python
#Compare product

c = df[df["Product"].isin(["Bread","Butter"])]

c= c.groupby("Product").agg(
    Total_Quantity =("Quantity","sum"),
    Total_sale = ("Unit_Price","sum"),
    Avg_rating = ("Rating","mean")
)
print(c)
```

```python
#10. Product report
pr = df.groupby("Product").agg(
    Total_Quantity =("Quantity","sum"),
    Total_sale = ("Unit_Price","sum"),
    Avg_rating = ("Rating","mean"),
    Avg_price = ("Unit_Price","mean"),
    Total_order = ("Order_ID","nunique"),
    Category = ("Category","first")
)
print(pr)
```

```python
#Phase:6 Time series Analysis
#1.Monthly sale
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
ms = df.groupby(df["Order_Date"].dt.to_period("M"))["Unit_Price"].sum()
print(ms)
```

```python
#2. Yearly sales
ms = df.groupby(df["Order_Date"].dt.year)["Unit_Price"].sum()
print(ms)
```

```python
#3.Quarterly Sales
ms = df.groupby(df["Order_Date"].dt.to_period("Q"))["Unit_Price"].sum()
print(ms)
```

```python
#4.Weekly Sales
ms = df.groupby(df["Order_Date"].dt.to_period("W"))["Unit_Price"].sum()
print(ms)
```

```python
#5.Growth Percentage
ms = df.groupby(df["Order_Date"].dt.to_period("M"))["Unit_Price"].sum()
growth_percentage = ms.pct_change()*100
print(growth_percentage)
```

```python
#6.Peak sales
ps = ms.sort_values(ascending=False).head(10)
print(ps)
```

```python
#7.Low sales 
ls = ms.sort_values().head(10)
print(ls)
```

```python
#8.Festival Trends

#fs =df.groupby("Festival")["Unit_Price"].sum().sort_values(ascending= False)
#print(fs)
```

```python
#9. Compare yearly growth
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
ys = df.groupby(df["Order_Date"].dt.year)["Unit_Price"].sum()

yg = ys.pct_change()*100
print(yg)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
mr = df.groupby(df["Order_Date"].dt.to_period("M"))["Unit_Price"].sum()
mr.columns = ["Month","ms"]
mr["growth_p"] = mr["ms"].pct_change()*100
print(mr)
```

```python
#phas:8 Advance Analytics
#1.Correlation between discount and sale
print("Correlation:")
print(df["Discount"].corr(df["Unit_Price"]))

#2.Covaroance between rating and sales
print("\nCovariance:")
print(df["Rating"].cov(df["Unit_Price"]))

```

```python
#3.Profit Margin
df["Profit"] = (df["Unit_Price"]- df["Discount"])*df["Quantity"]
df["Profit_Margin"] = (df["Profit"]/df["Unit_Price"])*100
print(df)
```

```python
#4. Growth rate
monthly_sale = df.groupby(df["Order_Date"].dt.to_period("M"))["Unit_Price"].sum()
growth_rate = monthly_sale.pct_change()*100
print(growth_rate)
```

```python
#5.Create KPI
total_sales = df["Unit_Price"].sum()
total_order = df["Order_ID"].nunique()
total_customer =df["Customer_ID"].nunique()
avg_order_value = df["Unit_Price"].mean()
avg_rating = df["Rating"].mean()
total_quantity = df["Quantity"].sum()


kpi = pd.DataFrame({
    "KPI":[
        "Total Sales",
        "Total Order",
        "Total Customer",
        "Average Order Value",
        "Average Rating",
        "Total Quantity"
    ],
    "Value":[
        total_sales,
        total_order,
        total_customer,
        avg_order_value,
        avg_rating,
        total_quantity
    ]
})
print(kpi)
```

```python
#6. Summary tables
#category wise 
summary = df.groupby("Category").agg(
    Total_Sales=("Unit_Price","sum"),
    Total_Quantity=("Quantity","sum"),
    AVG_Rating=("Rating","mean"),
    AVG_Price=("Unit_Price","mean")
)
print(summary)
```

```python
#7.Pivot Table
pivot_table = pd.pivot_table(
    df,
    values="Unit_Price",
    index="Category",
    columns="City",
    aggfunc="sum"
)
print(pivot_table)
```

```python
#8.Crosstab
cross_table = pd.crosstab(df["Category"],df["Payment_Method"],
                          values=df["Unit_Price"],
                          aggfunc="sum",
                          margins=True
                         )
print(cross_table)
```

```python
#9.Business insights
top_product = df.groupby("Product")["Quantity"].sum().idxmax()

top_city = df.groupby("City")["Unit_Price"].sum().idxmax()

top_category = df.groupby("Category")["Unit_Price"].sum().idxmax()


print("Top Product:",top_product)
print("Top City:",top_city)
print("Top Category",top_category)
```
