```python
#37.Calculate the Interquartile Range (IQR):
a = np.array(df["trip_duration"])
b = np.percentile(a,25)
c = np.percentile(a,75)
IQR = c - b
print(b)
print(c)
print("IQR:",IQR)
```

```python
#38.Find trips that are potential outliers using:

d = b - (1.5 * IQR)
e = c + (1.5 * IQR)

print("Lower outliers :",d)
print("Higher outliers :",e)


outlier = a[(a > d)&(a < e)]
print("Outliers:",outlier)

#39.Count the number of potential outlier trips.
print("Number of outliers:",len(outlier))





#40.Remove the potential outliers and calculate the new mean trip duration.
clean_data = a[(a >= d)&(a <= e)]

m = np.mean(clean_data)
print("New Mean :",m)
```
